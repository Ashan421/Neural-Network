1. What is Deep Learning? Briefly describe how it evolved and how it differs
from traditional machine learning.

 - Deep Learning is a specialized subset of Machine Learning that utilizes Artificial Neural Networks with multiple hidden layers to learn complex patterns and hierarchical representations directly from raw data.
 - Inspired by the structure and function of the human brain, these deep neural networks can automatically extract and select features from vast amounts of data, making them incredibly effective for complex tasks such as image recognition, natural language processing, and autonomous driving.

 - Evolution of deep learning is marked by several key periods:
    1. Early Foundations (1940s-1960s):
    - 1943: Warren McCulloch and Walter Pitts proposed the first mathematical model of a biological neuron, the McCulloch-Pitts Neuron.
    - 1957: Frank Rosenblatt developed the Perceptron, an algorithm capable of learning and making binary classifications, marking an early step in neural networks.
    - 1960s: Initial excitement waned due to the Perceptron's limitations and limited computational power, leading to reduced funding and research.

    2. The Resurgence (1980s-1990s):
    - 1986: The development of the Backpropagation algorithm enabled the efficient training of multi-layer neural networks by calculating gradients and adjusting weights to minimize error.
    - Late 1980s/Early 1990s: Researchers like Yann LeCun advanced Convolutional Neural Networks , leading to breakthroughs in tasks like handwritten digit recognition.

    3. The Modern Boom (2000s-Present):
    - Mid-2000s: Researchers like Geoffrey Hinton showed how to effectively train deeper networks layer by layer, leading to the popularization of the term "deep learning."
    - Key Enablers: The massive increase in data availability, the development of powerful GPUs for parallel processing, and continued algorithmic improvements finally provided the resources to train very deep and complex models efficiently.
    - 2012: AlexNet's landslide victory in the ImageNet image classification competition, significantly outperforming traditional methods, cemented Deep Learning's status as a revolutionary approach. Subsequent developments like Recurrent Neural Networks, Generative Adversarial Networks, and Transformers have driven the current AI revolution.

 - Deep learning is a subset of machine learning, but they differ fundamentally in their approach to data processing, especially feature engineering.
    1. Deep Learning
    - Automatic: The deep neural network automatically learns and extracts the hierarchical features directly from the raw data
    - Data Requirements: very large datasets. Performance continues to improve significantly as the data size increases.
    - Hardware: Requires specialized hardware, primarily GPUs or TPUs, for the computational intensity of training deep networks.
    - Problem Suitability: Best suited for complex tasks involving unstructured data like images, audio, and text.

    2. Traditional Machine Learning
    - Manual: A human expert must manually select, extract, and engineer the relevant features from the raw data
    - Data Requirements: Performs well with smaller and medium-sized datasets. Performance often plateaus after a certain data size.
    - Hardware:	Runs efficiently on standard CPUs.
    - Problem Suitability: Best suited for simpler, well-defined tasks, often with structured/tabular data.

2.  Explain the basic architecture and functioning of a Perceptron. What are its limitations?

 - The Perceptron is the simplest form of an Artificial Neural Network, serving as a fundamental supervised learning algorithm for binary classification. It was invented by Frank Rosenblatt in 1957, inspired by the structure of a biological neuron.

 - A single-layer Perceptron consists of:
    1. Input layer: Receives multiple input features x1 + x2 + ... + xn.
    2. Weights: Each input is assigned a weight w1 + w2 + ... + wn ,that indicates its importance.
    3. Summation function:
    - Computes the weighted sum of inputs: z = w1x1 + w2x2 + ... + wnxn + b.

 - The single-layer perceptron's major drawback, which led to the "AI Winter" in the 1970s, stems from its fundamentally linear nature:
    - Linearly Separable Data Only: The single-layer perceptron can only solve problems where the classes in the input data can be perfectly separated by a single straight line or a hyperplane.
    - The XOR Problem: Its most famous limitation is its inability to solve the Exclusive OR problem. The XOR function is a classic example of a non-linearly separable problem, meaning no single straight line can separate the inputs that result in an output of 1 from those that result in 0.
    - Binary Output: The output is strictly binary due to the step activation function, which prevents it from outputting probabilities or handling multi-class classification directly without modification.

3. Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.

 - A neural network node first computes a weighted sum of its inputs plus a bias. Without an activation function, every layer would simply perform a linear transformation. Stacking multiple linear transformations only results in another single linear transformation, meaning a deep network would be equivalent to a single-layer model.
 - The activation function, f(z), transforms this linear input into a non-linear output, which is then passed to the next layer.
    1. Introduce Non-Linearity: This is the primary role. Non-linearity allows the neural network to approximate any arbitrary function , enabling it to learn complex, non-linear mappings present in real-world data.
    2. Control Firing: It determines whether a neuron should be "activated" or "fired," passing the signal forward to the next layer.
    3. Ensure Differentiability: For deep learning models to be trained using backpropagation, the activation function must be differentiable to calculate the gradients needed to update the network's weights.

 - The three most common non-linear activation functions in deep learning are Sigmoid, Tanh , and ReLU.
    1. Sigmoid Function σ(x)
    - Properties:
      - Converts any real-valued number into a value between 0 and 1.
      - Often used for binary classification in the output layer.
    - Advantages:
      - Smooth gradient.
      - Easy to interpret as probability.
    - Disadvantages:
      - Vanishing gradient problem: for large |x|, gradient 0.
      - Not zero-centered: outputs are always positive.
      - Slower convergence in training.
    2. Hyperbolic Tangent (Tanh) Function
    - Properties:
      - Scales input between -1 and 1.
      - Used in hidden layers for centered activation.
    - Advantages:
      - Zero-centered → helps faster convergence.
      - Stronger gradients than Sigmoid in mid-range
    - Disadvantages:
      - Still suffers from vanishing gradient problem for large |x|.
    3. ReLU (Rectified Linear Unit) Function
    - Properties:
      - Introduces non-linearity without saturating for positive inputs.
      - Commonly used in hidden layers of deep neural networks.
    - Advantages:
      - Computationally efficient.
      - Faster convergence than Sigmoid or Tanh.
      - No vanishing gradient for positive x
    - Disadvantages:
      - Dying ReLU problem: neurons can “die” if inputs are always negative.

4. What is the difference between Loss function and Cost function in neural networks? Provide examples.

 - The difference between a Loss function and a Cost function in neural networks is primarily in the scope of their calculation, though the terms are often used interchangeably.
 - Both functions quantify the "error" or "badness" of a model's prediction, but they measure it at different levels.
    1. Loss Function
    - Scope: Calculates the error for a single training example.
    - Purpose: Measures the penalty for a single incorrect prediction.
    - Relationship: The Cost Function is typically the average of the individual Loss Functions over all $m$ training.
    - Examples of Loss Functions:
      - Mean Squared Error (MSE) for regression.
      - Binary Cross-Entropy Loss for binary classification.

    2. Cost Function
    - Scope: Calculates the average or total error across the entire training dataset or a mini-batch of data.
    - Purpose: Measures the overall performance of the model on the data and is the function that the optimization algorithm seeks to minimize.
    - Relationship: The Loss function is a component of the Cost function. The Cost function may also include a regularization term to prevent overfitting.
    - Examples of Cost Functions:
      - Mean Squared Error (MSE) Cost Function.
      - Cross-Entropy Cost Function.

5. What is the role of optimizers in neural networks? Compare Gradient Descent, Adam, and RMSprop.

 - Role of Optimizers: The process of training a neural network is an optimization problem: finding the set of weights and biases that results in the smallest possible error. Optimizers achieve this through Gradient Descent and its variants:
    1. Calculate the Gradient: The optimizer first uses backpropagation to calculate the gradient of the loss function with respect to every weight and bias in the network. This gradient indicates the direction of steepest ascent of the loss.
    2. Determine the Step: The optimizer then calculates the necessary update step for each parameter. It moves in the direction opposite the gradient to reduce the loss.
    3. Update Parameters: The parameters are updated by subtracting the calculated step size from their current values, iteratively moving the model toward the loss function's minimum.


    1. Gradient Descent (GD):
    - Mechanism: Uses the entire training dataset to calculate the gradient and update parameters in one single step.
    - Key Strengths: Mathematically guaranteed to converge to the minimum for convex functions.
    - Key Weaknesses: Very slow for large datasets. High memory usage.

    2. Adam or Adaptive Moment Estimation:
    - Mechanism: Combines Momentum and RMSprop to provide an adaptive learning rate for each parameter.
    - Key Strengths: Fast convergence and generally excellent performance across many tasks. Highly robust and often the default choice.
    - Key Weaknesses: Requires more memory to store the two moving averages.

    3. RMSprop or Root Mean Square Propagation:
    - Mechanism: Uses an exponentially decaying average of past squared gradients to adapt the learning rate, dividing the learning rate by this average.
    - Key Strengths: Effectively handles non-stationary objectives, making it good for Recurrent Neural Networks.
    - Key Weaknesses: Does not incorporate momentum, which can slow down convergence compared to Adam.

6. Write a Python program to implement a single-layer perceptron from
scratch using NumPy to solve the logical AND gate.

- Answer
        
      import numpy as np
      X = np.array([[0,0],[0,1],[1,0],[1,1]])
      y = np.array([0,0,0,1])

      np.random.seed(0)
      weights = np.random.randn(2)
      bias = 0.0
      lr = 0.1
      epochs = 20

      def step(x):
          return 1 if x >= 0 else 0

      for epoch in range(epochs):
      errors = 0
      for xi, target in zip(X, y):
              z = np.dot(weights, xi) + bias
              pred = step(z)
              update = lr * (target - pred)
              weights += update * xi
              bias += update
              errors += int(update != 0.0)
          if errors == 0:
              break

       import matplotlib.pyplot as plt

8. Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy.

 - Answer

       import tensorflow as tf
       from tensorflow.keras import layers, models
       from tensorflow.keras.utils import to_categorical

       # Load data
       (x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
       x_train = x_train.reshape(-1, 28*28).astype('float32')/255.0
       x_test = x_test.reshape(-1, 28*28).astype('float32')/255.0

       y_train_cat = to_categorical(y_train, 10)
       y_test_cat = to_categorical(y_test, 10)

       # Model
       model = models.Sequential([
          layers.Input(shape=(28*28,)),
          layers.Dense(256, activation='relu'),
          layers.Dense(128, activation='relu'),
          layers.Dense(10, activation='softmax')
       ])

       model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

       # Train
       history = model.fit(x_train, y_train_cat, epochs=5, batch_size=128, validation_split=0.1)

       # Print training accuracy (final epoch)
       train_acc = history.history['accuracy'][-1]
       print(f"Final training accuracy: {train_acc:.4f}")

- How to interpret training behavior:
    - If training accuracy increases while validation accuracy plateaus or drops to overfitting.
    - If both training and validation losses decrease smoothly to good fit.
    - If losses plateau early → learning rate may be too low or model capacity insufficient.
    - Look for gap between train and val curves to judge generalization.

    4. Handling class imbalance:
    - Resampling: oversample minority or undersample majority.
    - Class weights: pass class_weight to training to upweight fraud class.
    - Focal loss: reduces the impact of easy negatives.
    - Evaluation metrics: Precision, Recall, F1, PR-AUC are more informative than accuracy. Set operating point based on business needs.

    5. Optimizer & overfitting prevention:
    - Optimizer: Adam is a reasonable default. For production, consider SGD with momentum and learning rate scheduling if you need better generalization.
    - Prevent overfitting: dropout, L2 regularization, early stopping on validation metric, reduce model capacity, use cross-validation, and synthetic data augmentation for minority class if appropriate.
